```python
import pandas as pd
import random
```

```python
dataset=pd.read_csv('Book1.csv')
```

```python
print(dataset)
```

```python
cost=[]
for i in range(len(dataset)):
    cost.append(random.randint(1,15))
```

```python
search_map={}
i=0
while i<len(dataset):
    search_map[i] = [cost[i],[dataset.iloc[i,1],dataset.iloc[i,2], dataset.iloc[i,0]]]
    i=i+1
    for k in search_map:
        print(k, ' : ', search_map[k])
```
